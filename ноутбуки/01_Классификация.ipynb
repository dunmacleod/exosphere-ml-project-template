# Задание 1. Прогноз качества наблюдения

Компания «Экзосфера» хочет заранее оценивать риск непрохождения контроля качества наблюдения. Сохраните результаты исследования и выводы.

## SQL-запрос и выгрузка

Сформулируйте момент прогноза. Вставьте итоговый SQL как Markdown-блок `запросы`. Укажите единицу анализа, использованные таблицы, размер выгрузки и локальный путь к данным.

## Аудит данных, идентификатор и дубли

Загрузите CSV. Проверьте размер, типы, пропуски, распределения признаков, баланс классов и уникальность `session_id`. Найдите содержательные дубли без учёта идентификатора, объясните роль ID и обоснуйте обработку дублей до разделения.

## Разделение, инженерия признаков и предобработка

Самостоятельно выберите пропорции обучающей, валидационной и тестовой выборок. Выполните стратифицированное случайное разделение с фиксированным `random_state`. Проверьте отсутствие содержательных копий между выборками. Отложите тестовую выборку до итоговой оценки.

Исключите из модельных признаков идентификаторы и сведения, недоступные в момент прогноза. Создайте два или три обоснованных производных признака, доступных до съёмки. Организуйте заполнение пропусков, кодирование и масштабирование через `Pipeline` и `ColumnTransformer`: преобразования обучаются только на обучающих данных, в том числе внутри кросс-валидации.

## Сравнение моделей

Обучите `LogisticRegression` с обоснованными `penalty` и `C`, случайный лес или SVM с нелинейным ядром и один бустинг: XGBoost, LightGBM или CatBoost. Обоснуйте выбор алгоритмов и решение о весах классов.

Для каждой модели покажите на валидационной выборке Precision, Recall и F1 для `quality_failure=1`, а также матрицу ошибок. Объясните оба вида ошибок в контексте планирования наблюдений.

## Настройка лучшего алгоритма с Optuna

Выберите лучший алгоритм по F1 на валидационной выборке. Для практически равных результатов заранее объясните правило выбора с учётом сложности модели и требований бизнеса.

Настройте выбранный алгоритм с помощью Optuna и стратифицированной кросс-валидации на обучающей выборке. Ориентир — 20–30 испытаний. Обоснуйте метрику оптимизации, покажите пространство поиска, лучшие параметры и изменение качества. Тестовую выборку не используйте.

## Финальная оценка на тестовой выборке

Зафиксируйте финальную конфигурацию и один раз оцените её на тестовой выборке. Покажите Precision, Recall, F1 для класса `quality_failure=1` и матрицу ошибок. Итоговый F1 этого класса должен быть не ниже `0.75`.

## Интерпретация и рекомендация

После настройки выбранного алгоритма с помощью Optuna выполните SHAP-интерпретацию его итоговой настроенной модели. Укажите название модели и итоговый F1 на тестовой выборке.

Постройте глобальное объяснение — график суммарных эффектов SHAP. По графику:

- Определите, какие признаки вносят наибольший вклад в прогноз риска непрохождения контроля качества.
- Объясните, какие значения этих признаков связаны с повышением или снижением предсказываемого риска.
- Сопоставьте результат с логикой предметной области и проверьте, нет ли доминирования одного признака.

Дайте рекомендацию команде планирования: как использовать прогноз, какие ошибки остаются и где нужна дополнительная проверка.
