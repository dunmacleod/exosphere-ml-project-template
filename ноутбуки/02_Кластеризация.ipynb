# Задание 2. Сегментация коммерческих запросов

Компания «Экзосфера» хочет выделить профили коммерческого спроса. Сравните K-Means и DBSCAN, опишите группы и предложите применение сегментации.

## SQL-запрос и выгрузка

Подготовьте одну запись на `request_id`: количество и суммарную площадь территорий, числовые характеристики требований и контекстные сведения о клиенте и виде запроса. Самостоятельно выберите агрегацию и соединения. Проверьте единицу анализа, вставьте итоговый SQL как Markdown-блок `запросы` и укажите размер выгрузки и локальный путь к CSV.

## Подготовка признаков

Проверьте пропуски, дубли, распределения и выбросы. Выберите числовые признаки требований заказчиков и исключите идентификаторы. Обработайте пропуски, масштабируйте признаки и обоснуйте способ масштабирования. Сохраните исходные значения для профилирования.

Поля `client_type`, `industry_group`, `client_region` и `request_kind` используйте после кластеризации для интерпретации.

Кластеризацию K-Means и DBSCAN и расчёт их метрик выполняйте в полном пространстве масштабированных признаков.

## K-Means

Исследуйте диапазон числа кластеров методом локтя и Silhouette. Обоснуйте выбранное число кластеров метриками и содержанием групп. Покажите количество и долю запросов в каждом кластере.

## DBSCAN

Постройте график расстояний до k-го ближайшего соседа. Обоснуйте `eps` и `min_samples` с учётом размерности пространства. Покажите число кластеров и долю шума.

При наличии не менее двух нешумовых кластеров и допустимости расчёта рассчитайте Silhouette с исходными метками, включая `-1`, и объясните влияние шума. Если расчёт невозможен, укажите причину. Шумовые запросы опишите отдельно.

## Снижение размерности и визуализация

После кластеризации постройте двумерные проекции PCA и UMAP с раскраской по меткам каждого алгоритма. Шум DBSCAN обозначьте отдельно.

Интерпретируйте `explained_variance_ratio_` для PCA, зафиксируйте `random_state` для UMAP. Сравните проекции и объясните их ограничения.

## Сравнение и рекомендация

Сравните алгоритмы по метрикам, размерам групп, охвату, шуму и бизнес-смыслу. Обсудите зависимость от параметров и масштабирования; непроверенную устойчивость укажите как ограничение.

Для групп обоих алгоритмов покажите:

- Количество и долю запросов.
- Средние и медианные значения признаков в исходных единицах.
- Отличающие характеристики и контекстные категории.

Шумовые запросы опишите отдельно. Выберите один алгоритм, дайте группам содержательные описания и предложите применение сегментации при проектировании услуг и планировании ресурсов.
